# 📚 Parte 1 · Una lista dentro de otra: filas y columnas

🎬 Piense en la **sala de un cine**. Las sillas no están en una sola hilera: están en **filas**, y dentro de cada fila cada silla tiene su **número**. Para encontrar su silla usted hace dos cosas, siempre en el mismo orden: **primero busca la fila, después camina hasta la silla**.

En Python eso se guarda como **una lista cuyas casillas son listas**. Cada lista de adentro es una fila:

```python
sala = [
    ["L", "L", "X", "L"],     # fila 0
    ["X", "X", "L", "L"],     # fila 1
    ["L", "L", "L", "X"],     # fila 2
]
```

`"L"` es una silla libre y `"X"` una ocupada. Se escribe una fila por renglón para que **se vea** como un tablero, pero para Python es una sola lista con tres elementos (fíjese en la coma al final de cada fila).

| | columna 0 | columna 1 | columna 2 | columna 3 |
|---|---|---|---|---|
| **fila 0** | `sala[0][0]` | `sala[0][1]` | `sala[0][2]` | `sala[0][3]` |
| **fila 1** | `sala[1][0]` | `sala[1][1]` | `sala[1][2]` | `sala[1][3]` |
| **fila 2** | `sala[2][0]` | `sala[2][1]` | `sala[2][2]` | `sala[2][3]` |

### Se lee en dos pasos (igual que las fichas de la guía 5)
1. `sala[1]` → Python entrega **la fila 1 completa**: `["X", "X", "L", "L"]`. Es una lista.
2. A esa lista se le pide su posición 2: `sala[1][2]` → `"L"`.

Por eso van **dos pares de corchetes seguidos**: el primero escoge la **fila**, el segundo la **columna**. Si los invierte, lee otra casilla (o le sale un error).

### Lo que ya sabe sigue sirviendo
- `len(sala)` cuenta los elementos de la lista grande: **cuántas filas** hay.
- `len(sala[0])` cuenta los elementos de la fila 0: **cuántas columnas** hay.
- Las posiciones negativas también sirven: `sala[-1]` es la última fila y `sala[0][-1]` la última silla de la fila 0.
- Para **cambiar** una casilla se le asigna un valor, con sus dos posiciones: `sala[2][0] = "X"`.

````python
sala = [
    ["L", "L", "X", "L"],
    ["X", "X", "L", "L"],
    ["L", "L", "L", "X"],
]
print(len(sala))
print(len(sala[0]))
print(sala[1])
print(sala[1][2])
print(sala[0][-1])
sala[2][0] = "X"
print(sala[2])
````

**Qué pasó, línea por línea:**
- `len(sala)` dio `3`: la lista grande tiene tres elementos, que son las tres filas.
- `len(sala[0])` dio `4`: la fila 0 tiene cuatro sillas.
- `sala[1]` mostró la fila 1 **entera**, con sus corchetes: es una lista.
- `sala[1][2]` mostró una sola silla: la de la fila 1, columna 2.
- `sala[0][-1]` leyó la última silla de la fila 0.
- `sala[2][0] = "X"` ocupó la primera silla de la fila 2; por eso al mostrar `sala[2]` la primera casilla ya es `X`.

> ⚠️ **El error de hoy:** `IndexError: list index out of range`. Aparece cuando pide una fila o una columna que no existe. En esta sala las filas van de **0 a 2** y las columnas de **0 a 3**: `sala[3][0]` y `sala[0][4]` dan error.

# 📚 Parte 2 · Recorrer el tablero: un ciclo dentro de otro

Para revisar **todas** las sillas de un cine, el acomodador hace esto: toma una **fila** y la camina silla por silla; cuando termina, pasa a la **siguiente fila** y vuelve a empezar desde la primera silla. Son dos recorridos, **uno dentro del otro**.

### Receta a) Fila por fila
Un solo `for`. En cada vuelta, la variable recibe **una fila completa** (una lista). Sirve para **mostrar** el tablero:

```python
for fila in sala:
    print(" ".join(fila))
```

`" ".join(fila)` es lo de la guía 3: une los textos de la lista con un espacio en medio. (Solo sirve si las casillas son **textos**).

### Receta b) Casilla por casilla
Dos `for`. El de **afuera** toma una fila; el de **adentro** recorre las casillas de esa fila. Sirve para **contar, sumar o buscar un dato**:

```python
for fila in sala:            # afuera: una fila a la vez
    for silla in fila:       # adentro: cada silla de esa fila
        ...                  # aquí se revisa UNA silla
```

Con 3 filas de 4 sillas, lo que está dentro del segundo `for` se repite **3 × 4 = 12 veces**. La **sangría** dice qué pertenece a cada ciclo: lo que está al nivel del `for` de adentro pasa **una vez por fila**; lo que está más adentro pasa **una vez por silla**.

### Receta c) Con posiciones
Cuando hay que decir **dónde** está algo, o **cambiar** la casilla, se necesitan los números de fila y columna. Se usa `range(len())`, como en la guía 1, pero dos veces:

```python
for f in range(len(sala)):             # f = 0, 1, 2
    for c in range(len(sala[f])):      # c = 0, 1, 2, 3
        ... sala[f][c] ...
```

`len(sala[f])` es la cantidad de sillas **de esa fila**.

````python
sala = [
    ["L", "L", "X", "L"],
    ["X", "X", "L", "L"],
    ["L", "L", "L", "X"],
]
for fila in sala:
    print(" ".join(fila))
print("---")
libres = 0
for fila in sala:
    for silla in fila:
        if silla == "L":
            libres = libres + 1
print("Libres:", libres)
print("---")
for f in range(len(sala)):
    for c in range(len(sala[f])):
        if sala[f][c] == "X":
            print("Ocupada: fila", f, "columna", c)
````

**Qué pasó, por bloques:**
- **Primer bloque:** un `for`, tres vueltas, una por fila. En cada una `" ".join(fila)` armó el renglón. Así se «dibuja» el tablero.
- **Segundo bloque:** es el **patrón de contar** de la guía 2 (contador en cero **antes**, `if` **dentro**, `print` **después**). Lo único nuevo es que hay dos ciclos, para llegar a cada silla. El `if` se hizo 12 veces.
- **Tercer bloque:** como había que decir **en qué fila y columna** está cada silla ocupada, se recorrió con posiciones. `sala[f][c]` es la silla de la vuelta actual.

> 🔑 **Regla para escoger:** ¿solo necesito el **dato**? → receta b. ¿Necesito la **posición** o voy a **cambiar** la casilla? → receta c.

# 📚 Parte 3 · Sumar por filas, sumar por columnas y buscar una posición

Una tabla de números se puede sumar **a lo ancho** (por filas) o **a lo largo** (por columnas). Piense en una planilla de ventas: cada **fila** es una tienda y cada **columna** es un día.

### Sumar una fila
La fila ya es una lista, así que es el **patrón de acumular** de siempre, sobre `ventas[f]`.

### Sumar una columna (receta d)
Una columna **no** es una lista: sus datos están repartidos, uno en cada fila. Para recorrerla se deja **quieta la columna** y se **mueve la fila**:

```python
total = 0
for f in range(len(ventas)):      # f recorre las filas
    total = total + ventas[f][2]  # siempre la columna 2
```

Para sumar **todas** las columnas, el ciclo de **afuera** recorre las columnas y el de **adentro** las filas. Es la receta c con los ciclos intercambiados.

### Buscar y decir dónde está
Es el **patrón de buscar** dentro de una función, como `buscar()` de la guía 5, con dos novedades:
- Se recorre **con posiciones** (receta c), porque lo que se quiere es la posición.
- Se devuelven **dos datos**: `return f, c`. Como vio el martes, eso entrega una **tupla** `(fila, columna)`.
- Si se recorre todo y no aparece, después de los dos ciclos va `return None`.

Recuerde de la guía 6: `return` **termina la función en ese momento**. Por eso entrega la **primera** casilla que encuentre y no sigue buscando.

````python
ventas = [
    [10, 20, 30],
    [5, 15, 25],
]
for f in range(len(ventas)):
    total = 0
    for valor in ventas[f]:
        total = total + valor
    print("Fila", f, "suma", total)

for c in range(len(ventas[0])):
    total = 0
    for f in range(len(ventas)):
        total = total + ventas[f][c]
    print("Columna", c, "suma", total)

def buscar(tabla, dato):
    for f in range(len(tabla)):
        for c in range(len(tabla[f])):
            if tabla[f][c] == dato:
                return f, c
    return None

print(buscar(ventas, 15))
print(buscar(ventas, 99))
````

**Qué pasó, por bloques:**
- **Por filas:** el acumulador `total` vuelve a cero al empezar cada fila (igual que el contador del ejercicio 5). Salieron dos renglones porque hay dos filas.
- **Por columnas:** ahora el ciclo de afuera es `c` y el de adentro es `f`. `ventas[f][c]` se sigue escribiendo **fila primero, columna después**: eso nunca cambia, aunque los ciclos estén intercambiados. Salieron tres renglones porque hay tres columnas.
- **`buscar(ventas, 15)`** encontró el 15 en la fila 1, columna 1, y devolvió la tupla `(1, 1)`.
- **`buscar(ventas, 99)`** recorrió las seis casillas, no entró nunca al `if`, salió de los dos ciclos y llegó al `return None`.

# 📚 Parte 4 · Crear un tablero vacío y validar una posición

### Crear el tablero (receta e)
Un tablero grande no se escribe a mano: se **arma con ciclos**. Es el patrón de **armar una lista nueva** de la guía 2, hecho dos veces: se arma una fila casilla por casilla, y esa fila se agrega al tablero.

```python
tablero = []
for f in range(3):            # tres filas
    fila = []                 # una fila NUEVA en cada vuelta
    for c in range(3):        # tres casillas
        fila.append("-")
    tablero.append(fila)
```

### ⚠️ La trampa de la fila repetida
Parece más corto hacer una fila y ponerla tres veces. **No sirve:**

```python
fila = ["-", "-", "-"]
tablero = [fila, fila, fila]     # ❌ no son tres filas: es UNA, anotada tres veces
```

Es lo mismo que pasó en la guía 1 cuando una lista se «copió» con `=`: los tres puestos del tablero señalan **la misma lista**. Si marca una casilla, la ve cambiar en las tres filas. Por eso `fila = []` va **dentro** del ciclo de afuera: así cada vuelta estrena una lista distinta.

### Validar antes de usar
Si el usuario escribe una fila que no existe, el programa se cae con `IndexError`. Antes de leer o cambiar `tablero[f][c]` se pregunta si la posición **está dentro**:

```python
if f >= 0 and f < len(tablero) and c >= 0 and c < len(tablero[0]):
    ...   # sí existe
```

Hay que revisar también que **no sea negativa**: `tablero[-1][0]` no da error, lee la **última** fila, y el programa marcaría una casilla que el usuario no pidió.

Y un detalle con las personas: ellas cuentan desde **1**. Si el usuario dice «fila 1», en Python es la fila `0`: se le **resta 1** a lo que escribe.

````python
def crear_tablero(filas, columnas, relleno):
    tablero = []
    for f in range(filas):
        fila = []
        for c in range(columnas):
            fila.append(relleno)
        tablero.append(fila)
    return tablero

bueno = crear_tablero(2, 3, "-")
bueno[0][0] = "X"
print(bueno)

fila = ["-", "-", "-"]
malo = [fila, fila]
malo[0][0] = "X"
print(malo)

def existe(tablero, f, c):
    return f >= 0 and f < len(tablero) and c >= 0 and c < len(tablero[0])

print(existe(bueno, 1, 2))
print(existe(bueno, 2, 0))
print(existe(bueno, -1, 0))
````

**Qué pasó:**
- En `bueno`, la `X` quedó **solo** en la fila 0: cada fila es una lista distinta.
- En `malo`, la `X` apareció en **las dos filas**, aunque solo se marcó `malo[0][0]`: las dos filas son la misma lista.
- `existe()` es una función que responde sí o no (guía 6): devuelve directamente el resultado de la condición. La posición `(1, 2)` existe; la fila `2` no (solo hay 0 y 1); la fila `-1` tampoco se acepta.

### Ejercicio 9 · El tablero con coordenadas  
**Nivel:** ⭐⭐⭐ Reto · **Tipo:** modificar un programa que funciona  
**👥 En pareja:** A pilotea (escribe), B copilotea (dicta y revisa)

**📖 La situación:** la función `mostrar()` del ejercicio 4 ya funciona, pero para jugar hace falta saber **qué número tiene cada fila y cada columna**. Hay que **modificarla** para que muestre los números, contando **desde 1**, como cuentan las personas.

**📝 Lo que debe hacer, paso a paso:**
1. Ejecute la celda tal como viene, para ver qué hace hoy.
2. En papel: escriba cómo debe quedar la salida y marque qué es nuevo (un renglón arriba y un número al comienzo de cada fila).
3. **Renglón de arriba:** arme una lista con los números de columna **como textos** (`"1"`, `"2"`, `"3"`), con un ciclo y `.append()`. Recuerde que `.join()` solo une textos: convierta cada número con `str()`. Muéstrelo con **dos espacios** adelante para que quede alineado.
4. **Cada fila:** ahora necesita el número de la fila, así que el ciclo debe recorrer **con posiciones**. Muestre el número de la fila, un espacio y la fila unida.
5. La función debe servir para un tablero de **cualquier tamaño**: no escriba el 3 a mano, use `len()`.

**👀 Así debe verse al ejecutarlo**:
```
1 2 3
1 X O -
2 - X -
3 O - X
  1 2 3 4
1 L L X L
2 X L L L
```

<details><summary>💡 Ver pista (último recurso: primero intente y relea la explicación)</summary>

El ciclo de las columnas es `for c in range(len(tablero[0])):` y lo que se agrega a la lista es `str(c + 1)`. Para las filas: `for f in range(len(tablero)):` y la fila es `tablero[f]`.

</details>

````python
def mostrar(tablero):
    for fila in tablero:
        print(" ".join(fila))

mostrar([["X", "O", "-"], ["-", "X", "-"], ["O", "-", "X"]])
mostrar([["L", "L", "X", "L"], ["X", "L", "L", "L"]])
````
